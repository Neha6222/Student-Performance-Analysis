```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.DataFrame({
    'students':['Neha','priya','pinky','sonu','pandu'],
    'Maths':[100,90,90,80,70],
    'Science':[98,99,97,56,89],
    'Social':[96,56,49,79,80]
})
#df.head()
#df.info()
#df.describe()
df['Total']=df['Maths']+df['Science']+df['Social']
df['Average']=df['Total']/3
df['Rank']=df['Total'].rank(ascending=False)
print(df)
```

```python
df.set_index('students')['Average'].plot(kind='bar',color='green')
plt.title("Average marks of students")
plt.ylabel('Average marks')
plt.show()
```

```python
plt1=df[['Maths','Science','Social']].mean()
plt1.plot(kind='bar',color=['pink','red','yellow'])
plt.title("Average marks of students")
plt.show()
```

```python
subjects=df[['Maths','Science','Social']]
corr=subjects.corr()
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
data=df[['Maths','Science','Social']]
sns.boxplot(data)
plt.title("Distribution of marks")
plt.show()
```
